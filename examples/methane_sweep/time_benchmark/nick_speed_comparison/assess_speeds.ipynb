```python
import hippynn
import torch
```

```python
all_data = torch.load('./speed_eval.pt')
n_configs = metrics = None
```

```python
metrics = all_data.pop('metrics', metrics)
n_configs = all_data.pop('n_configs',n_configs)
```

```python
batch_size=1024
all_min = {}
for model,speeds in metrics.items():
    model = dict(model)
    if model['batch_size']!=batch_size:
        continue
    model.pop('batch_size')
    min_speed = min(speeds)
    
    model = tuple(sorted(model.items(),key=lambda kv:kv[0]))
    all_min[model] = min_speed
    
```
